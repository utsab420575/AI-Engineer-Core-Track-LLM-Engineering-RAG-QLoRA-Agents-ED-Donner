```python
# Import OpenAI library
# WHY: We use this library to communicate with AI models
# through an API. Groq supports the OpenAI-compatible client.
from openai import OpenAI


# Import os module
# WHY: It helps us access environment variables,
# such as our secret Groq API key.
import os


# Import load_dotenv function
# WHY: It allows Python to read secret keys
# stored in a .env file.
from dotenv import load_dotenv
```

```python
# Load environment variables from the .env file
# WHY: Without this, the API key stored in .env
# may not be available through os.environ.
load_dotenv()


# Get the Groq API key from environment variables
# WHY: We need this key to prove our identity
# and get permission to use Groq's API.
api_key = os.getenv("GROQ_API_KEY")


# Check whether the API key exists
# WHY: If the key is missing, the API connection
# will fail. This gives us a clear error message.
if not api_key:
    raise ValueError("GROQ_API_KEY not found in .env file")


# Create a client to connect to Groq
# WHY: The client acts as a bridge between
# our Python program and Groq's AI service.
client = OpenAI(

    # Pass our secret API key
    # WHY: Groq uses this key to authenticate our request.
    api_key=api_key,

    # Set Groq's API server address
    # WHY: The OpenAI library normally connects to OpenAI.
    # Changing this URL makes it send requests to Groq instead.
    base_url="https://api.groq.com/openai/v1"
)


# Send a question to the AI model
# WHY: This requests the AI to generate an answer
# based on the question we provide.
response = client.responses.create(

    # Input is the question or instruction for the AI
    # WHY: This tells the model what we want it to do.
    input="Explain the importance of fast language models",

    # Specify which AI model should answer
    # WHY: Different models have different capabilities.
    # Here we select Groq's hosted GPT-OSS 20B model.
    model="openai/gpt-oss-20b"
)


# Print a heading in the terminal
# WHY: This makes the output easier to read.
# \n means start a new line.
print("\nGroq Response:\n")


# Display the AI-generated answer
# WHY: response contains the API result.
# output_text extracts the generated text from it.
# print() displays that text on the screen.
print(response.output_text)
```
