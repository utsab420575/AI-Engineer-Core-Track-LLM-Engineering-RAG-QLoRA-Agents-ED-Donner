```python
from openai import OpenAI
import os
from dotenv import load_dotenv
load_dotenv()
api_key = os.environ.get("GROQ_API_KEY")
if not api_key:
    raise ValueError("GROQ_API_KEY not found in .env file")


client = OpenAI(

    # Pass our secret API key
    # WHY: Groq uses this key to authenticate our request.
    api_key=api_key,

    # Set Groq's API server address
    # WHY: The OpenAI library normally connects to OpenAI.
    # Changing this URL makes it send requests to Groq instead.
    base_url="https://api.groq.com/openai/v1"
)

# Define the system prompt
# WHY: Tells AI its role, tone and behaviour.

system_prompt = """
You are a snarky assistant that analyzes website content.
Provide a short, humorous summary.
Ignore navigation-related text.
Respond in Markdown.
"""


# Define the user prompt
# WHY: Tells AI what task to perform.

user_prompt = """
Explain the importance of fast language models.
"""


# Create a list of messages
# WHY: We send both system and user instructions
# together in one API request.

messages = [
    {
        "role": "system",
        "content": system_prompt
    },
    {
        "role": "user",
        "content": user_prompt
    }
]


# The most important difference to remember:
#  client.responses.create() → Send input, retrieve output_text.
        # response = client.responses.create(

        #     input="Explain the importance of fast language models",
        #     model="openai/gpt-oss-20b"
        # )

#  client.chat.completions.create() → Send messages, retrieve choices[0].message.content.
# Both methods send requests to the AI model and return generated responses, 
#                       but they use different API interfaces and response structures.


# Send messages to Groq using the Chat Completions API.
# WHY: This method sends a conversation to the AI model
# and asks it to generate a response.

response = client.chat.completions.create(

    # Specify which AI model should process our messages.
    # WHY: Groq provides multiple AI models.
    # Here we are selecting the GPT-OSS 20B model.
    model="openai/gpt-oss-20b",

    # Send the conversation messages to the AI.
    # WHY: The messages variable contains our system prompt
    # and user prompt.
    messages=messages
)


# Extract the AI-generated response.
# WHY:
# response = Contains the complete API response.
# choices = Contains the generated response options.
# [0] = Selects the first response.
# message = Accesses the message generated by the AI.
# content = Extracts the actual text of the answer.

print(f"Response from First Message: {response.choices[0].message.content}")



#another message we will send to the model
messages=[{
    "role": "system",
    "content": "you are a helpful assistant"
},{
    "role": "user",
    "content": "what is 2+2"
}]

response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=messages
)
print(f"\n \n Response from Second Message: {response.choices[0].message.content}")
```
