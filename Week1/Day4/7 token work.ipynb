```python

# Import the tiktoken library.
# tiktoken is used to count and encode text into tokens
# for OpenAI language models.
import tiktoken

# Select the token encoding method.
# "o200k_base" is used by several recent OpenAI models.
# Since a model-specific encoding may not be available,
# we use "o200k_base" as an approximation.
encoding = tiktoken.get_encoding("o200k_base")

# Define the text that we want to tokenize.
text = "Hi my name is Ed and I like banoffee pie"

# Convert the text into tokens.
# Each token is represented by a unique integer ID.
tokens = encoding.encode(text)


print(f"Text: {text}")
#print the list of tokens generated from the text
print(f"Tokens: {tokens}")
```

```python
# Loop through each token ID in the tokens list
for token_id in tokens:

    # Convert the current token ID (integer) into readable text
    # [token_id] makes it a list because decode() expects a list
    token_text = encoding.decode([token_id])

    # Print the token ID and its corresponding text
    # f-string inserts the values of the variables
    print(f"{token_id} = {token_text}")
```

```python
# Decode all token IDs together
token_text = encoding.decode(tokens)

# Print the complete text
print(token_text)
```

```python
import os
from dotenv import load_dotenv
load_dotenv()  # Load environment variables from a .env file
api_key = os.getenv("GROQ_API_KEY")  # Get the API key from environment variables
if not api_key:
    raise ValueError("GROQ_API_KEY environment variable is not set. Please set it in your .env file.")
elif not api_key.startswith("gsk"):
    raise ValueError("GROQ_API_KEY does not appear to be a valid API key. Please check your .env file.")
else:
    print("GROQ_API_KEY is set correctly.") 
```

```python
from openai import OpenAI
openai_client = OpenAI(api_key=api_key,
                       base_url="https://api.groq.com/openai/v1")

# Generate response
response = openai_client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {"role": "system", "content": "You are a helpful assistant"},
        {"role": "user", "content": "Hi! I'm Ed!"}
    ]
)
```

```python
response_message = response.choices[0].message.content
print(response_message)
```
