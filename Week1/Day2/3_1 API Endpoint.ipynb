```python
# Import required libraries
import os
from dotenv import load_dotenv

#load_dotenv() → Loads variables from the .env file into the environment.
# Load environment variables from the .env file.
# override=True replaces existing environment values.
load_dotenv(override=True)

# Retrieve the OpenAI API key from environment variables.
#os.getenv() → Reads a variable from the environment.
api_key = os.getenv("GROK_API_KEY")

# Check whether the API key exists and has the expected format.
if not api_key:
    print("No API key was found - please head over to the troubleshooting notebook in this folder to identify & fix!")

else:
    print("API key found and looks good so far!")
```

```python

# --------------------------------------------------
# STEP 1: IMPORT THE REQUESTS LIBRARY
# --------------------------------------------------

# Import the requests library.
# WHY:
# It allows Python to send HTTP requests
# to external APIs and communicate with web servers.
#
# We will use it later to send our API request
# to the OpenAI API.

import requests


# --------------------------------------------------
# STEP 2: PREPARE THE API REQUEST HEADERS
# --------------------------------------------------

# Create a dictionary containing the HTTP request headers.
#
# Authorization:
#   Sends our API key to the server for authentication.
#
# f"Bearer {api_key}":
#   Inserts our API key into the Authorization header.
#   The word "Bearer" tells the server that we are
#   using a Bearer Token for authentication.
#
# Content-Type:
#   Specifies the format of the data we are sending.
#
# application/json:
#   Tells the server that our request body
#   contains JSON-formatted data.
#
# WHY:
# The API needs to know who is making the request
# and what format the submitted data uses.

headers = {
    "Authorization": f"Bearer {api_key}",
    "Content-Type": "application/json"
}


# --------------------------------------------------
# STEP 3: PREPARE THE API REQUEST PAYLOAD
# --------------------------------------------------

# Create a dictionary containing the information
# we want to send to the AI model.
#
# A payload is the data included in an API request.
#
# WHY:
# We need to tell the API which model to use,
# how to configure the request, and what question
# we want the AI model to answer.

payload = {

    # Specify the AI model we want to use.
    # The API will attempt to generate a response
    # using this model.
    "model": "openai/gpt-oss-20b",

    # Specify the reasoning effort for the model.
    # "none" means no additional reasoning effort
    # is requested through this parameter.
    "reasoning_effort": "high",

    # Provide the conversation messages.
    #
    # messages is a list of dictionaries.
    # Each dictionary represents one message
    # in the conversation.

    "messages": [

        # Define the user's message.
        #
        # role:
        #   Identifies who is sending the message.
        #
        # "user":
        #   Means this message comes from the user.
        #
        # content:
        #   Contains the actual text or instruction
        #   that we want the AI model to process.

        {
            "role": "user",
            "content": "Tell me a fun fact"
        }
    ]
}

payload
```

# Send a POST request to the OpenAI API with headers and payload.
response = requests.post(
    "https://api.openai.com/v1/chat/completions",
    headers=headers,  # Send API key and content type.
    json=payload     # Send the request data in JSON format.
)

# Display the complete API response as a Python dictionary.
response.json()

```python

# --------------------------------------------------
# STEP 5: SEND THE API REQUEST TO OPENAI
# --------------------------------------------------

# Use requests.post() to send an HTTP POST request
# to the OpenAI API endpoint.

# POST:
# An HTTP method used to send data to a server.
#
# API URL:
# https://api.openai.com/v1/chat/completions
#
# This is the Chat Completions API endpoint.

response = requests.post(

    # Specify the OpenAI API endpoint.
    # This is where our request will be sent.
    "https://api.groq.com/openai/v1/chat/completions",

    # Send the authentication headers.
    #
    # headers=headers:
    # Uses the headers dictionary we created earlier.
    # It contains our API key and content type.
    #
    # This allows OpenAI to authenticate our request
    # and understand that we are sending JSON data.

    headers=headers,

    # Send the request payload as JSON.
    #
    # json=payload:
    # Converts our Python dictionary into JSON format
    # and includes it in the HTTP request body.
    #
    # The payload contains:
    # 1. The AI model we want to use.
    # 2. The reasoning effort setting.
    # 3. The user's message.
    #
    # OpenAI uses this information to generate
    # the requested response.

    json=payload
)


# --------------------------------------------------
# STEP 6: EXTRACT THE API RESPONSE AS JSON
# --------------------------------------------------

# Convert the server's response into a Python dictionary.
#
# response:
#   Contains the complete HTTP response returned by OpenAI.
#
# .json():
#   Converts the JSON response body into Python objects,
#   usually a dictionary containing response information.
#
# WHY:
# We need to read the information returned by the API,
# including the AI-generated answer and other metadata.
#
# IMPORTANT:
# This displays the complete response.
# It does not extract only the AI-generated text.

response.json()
```
