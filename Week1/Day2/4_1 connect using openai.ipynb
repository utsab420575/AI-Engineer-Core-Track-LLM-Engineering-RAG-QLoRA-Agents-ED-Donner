```python

import os
from dotenv import load_dotenv
from openai import OpenAI

# Load environment variables
load_dotenv(override=True)

# Get Grok API key
api_key = os.getenv("GROK_API_KEY")

# Check API key
if not api_key:
    raise ValueError("GROK_API_KEY is not set.")

print("GROK_API_KEY is set.")


#client = OpenAI()
# Create an OpenAI client instance.
# This initializes the API client, automatically reads the
# OPENAI_API_KEY environment variable, and prepares the client
# to send requests to OpenAI's API.

#But we use Grok API endpoint instead of OpenAI's default endpoint. So we need to specify the base_url parameter when creating the client instance.
# Create Grok API client
client = OpenAI(
    api_key=api_key,
    base_url="https://api.groq.com/openai/v1"
)

# Generate response
response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    messages=[
        {
            "role": "user",
            "content": "Write a short poem about the beauty of nature."
        }
    ]
)

# Display AI response
print(response.choices[0].message.content)
```

With Comments 

```python

# --------------------------------------------------
# STEP 1: IMPORT REQUIRED LIBRARIES
# --------------------------------------------------

import os
# WHY:
# Provides access to environment variables.
# We use it to retrieve our API key securely.

from dotenv import load_dotenv
# WHY:
# Imports load_dotenv() to load environment variables
# stored in a .env file.

from openai import OpenAI
# WHY:
# Imports the OpenAI client class.
# We can also use this SDK to communicate with
# other providers that support the OpenAI-compatible API.


# --------------------------------------------------
# STEP 2: LOAD ENVIRONMENT VARIABLES
# --------------------------------------------------

load_dotenv(override=True)
# WHY:
# Loads environment variables from the .env file.
# override=True means values from the .env file
# will override existing environment variables
# with the same names.


# --------------------------------------------------
# STEP 3: RETRIEVE GROQ API KEY
# --------------------------------------------------

api_key = os.getenv("GROK_API_KEY")
# WHY:
# Retrieves the API key stored in the environment
# variable named GROK_API_KEY.
# The variable name must match the name in .env.


# --------------------------------------------------
# STEP 4: CHECK API KEY
# --------------------------------------------------

if not api_key:
    raise ValueError("GROK_API_KEY is not set.")
else:
    print("GROK_API_KEY is set.")

# --------------------------------------------------
# STEP 5: CREATE GROQ API CLIENT
# --------------------------------------------------

# Normally, OpenAI() uses OpenAI's default API endpoint.
# However, we are using Groq's API instead.
# Therefore, we provide the Groq API key and custom base URL.

client = OpenAI(
    api_key=api_key,
    # WHY:
    # Provides the API key for authentication.

    base_url="https://api.groq.com/openai/v1"
    # WHY:
    # Specifies Groq's OpenAI-compatible API endpoint.
    # This directs API requests to Groq instead of OpenAI.
)


# --------------------------------------------------
# STEP 6: SEND REQUEST TO THE AI MODEL
# --------------------------------------------------

response = client.chat.completions.create(
    model="openai/gpt-oss-20b",
    # WHY:
    # Specifies the AI model hosted by Groq.
    # This model processes the user's prompt.

    messages=[
        {
            "role": "user",
            # WHY:
            # Identifies the message as coming from the user.

            "content": "Write a short poem about the beauty of nature."
            # WHY:
            # Contains the actual prompt or instruction
            # that we want the AI model to process.
        }
    ]
)


# --------------------------------------------------
# STEP 7: DISPLAY THE AI-GENERATED RESPONSE
# --------------------------------------------------

print(response.choices[0].message.content)
```
